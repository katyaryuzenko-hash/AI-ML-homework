## Домашнє завдання: Цикл for

### Task1 : Цикл for - базовий

```python
for n in range (1, 8):
   print(n)
```

### Task2 : Цикл for по списку

```python
names = ["Олег", "Анна", "Іван", "Марія"] 

for name in names: 
    print(f"Привіт, {name}!")
```

### Task3 : Цикл з enumerate

```python
grades = [85, 92, 78, 95, 88]

for i, grade in enumerate(grades):
    print(f"Студент {i+1}: {grade}")
```

### Task4 : Сума парних чисел

```python
total = 0

for x in range (1,21):
    if x % 2 == 0:
        total = total + x

print(f"Сума:", total)
```
