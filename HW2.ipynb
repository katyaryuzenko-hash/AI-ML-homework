### Task1 - Списки - базові операції

```python
# Cписок температур за тиждень (у градусах Цельсія)

temperatures = [22, 25, 19, 23, 26, 20, 21, 22]
print ('All_temp:', temperatures)
```

```python
print('First element:', temperatures[0]) #Виведення першого елементу
print('Last temp:', temperatures[-1]) #Виведення останнього елементу
```

```python
#Додавання температури 24

temperatures.append(24)
print ('All_temp:', temperatures)
```

```python
#Видалення температури 19

temperatures.remove(19)
print ('All_temp:', temperatures)
```

```python
print ('Length of a list:', len(temperatures)) # Довжина списку
print ('Sum temperatures:', sum(temperatures)) # Обчислення сум всіх температур
print ('Max temp:', max(temperatures)) # Максимальна температура
```

### Task2 - Словники - базові операції

```python
# Crete dictionary

dict_about_me = {
    "name" : "Kateryna",
    "age" : 29,
    "city" : "Kyiv"
}
```

```python
# Виведення значення ключа "name"

print(dict_about_me["name"])
```

```python
#Зміна значення ключа "age" на 21

dict_about_me["age"] = 21
print(f"Changed age:" , dict_about_me["age"])
```

```python
#Додавання нового ключа "hobby" зі значенням "programming"

dict_about_me["hobby"] = "programming"
print(f"Hobby:" , dict_about_me["hobby"])
```

```python
#Виведення всих ключів та значень

print(dict_about_me.keys())
print(dict_about_me.values())
```
