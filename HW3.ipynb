## Домашнє завдання: Умовні конструкції (if-elif-else)

### Task1 : Основи роботи з умовними конструкціями (if-elif-else)

```python
cooking_time = 5 # (час варіння яєць у хвилинах)

if cooking_time < 4: 
    print("Яйце рідке")
elif 4 <= cooking_time < 7:
    print("Яйце всмятку")
elif 7 <= cooking_time < 10:
    print("Яйце вкруту")
else:
    print("Яйце переварене")
```

### Task2 : Перевірка оцінки

```python
grade = 72

if grade >= 90:
    print("Відмінно")
elif 75 <= grade < 90:
    print("Добре")
elif 60 <= grade < 75:
    print("Задовільно")
else:
    print("Незадовільно")
```

### Task3 : Перевірка на наявність

```python
products = ["молоко", "яйця", "масло", "сир"]
```

```python
#Чи є "яйця" в списку

find_products = "яйця"
if find_products in products:
  print (f"{find_products} є в холодильнику")
else: 
  print(f"{find_products} не знайдено")
```

```python
#Чи є "хліб" в списку

find_products = "хліб"
if find_products in products:
    print(f"{find_products} є в холодильнику")
else:
    print(f"{find_products} не знайдено")
```
