## Домашнє завдання: Цикл while. List comprehensions

### Task 1 : Цикл while

```python
counter = -3

while counter <= 5:
    counter += 1
    print(counter)
```

### Task 2 : List comprehension - прості операції

```python
# Квадрати всіх чисел
numbers = [11, 4, 9, 16, 7, 5]

result1 = [n * n for n in numbers]

print(result1)
```

```python
# Тільки парні числа
numbers = [11, 4, 9, 16, 7, 5]

result2 = [n for n in numbers if n % 2 == 0]

print (result2)
```

```python
# Всі числа помножені на 10
numbers = [11, 4, 9, 16, 7, 5]

result3 = [n * 10 for n in numbers ]

print(result3)
```

### Task 3 : List comprehension - комбінована логіка

```python
# Написала розгорнуто, щоби зберегти 2 варіанти, для легшого розуміння на майбутнє собі
numbers = list(range(1, 11))

changed_list = []
for n in numbers:
    if n % 2 == 0:
        changed_list.append(n*2)
    else:
        changed_list.append(n - 1)
print(changed_list)
```

```python
numbers = list(range(1, 11))

changed_list = [n * 2 if n % 2 == 0 else n - 1 for n in numbers]

print(changed_list)
```
