### Task2 - Cтворюємо арифметичні операції

```python
var1 = 15
var2 = 4

print(var1 + var2)
print(var1 - var2)
print(var1 * var2)
print(var1 / var2)
print(var1 // var2)
print(var1 % var2)
print(var1 ** var2)
```
